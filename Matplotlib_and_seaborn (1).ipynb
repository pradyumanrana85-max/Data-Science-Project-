```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("IPL2025Batters.csv")
```

```python
# Q1 What is the distribution of batting averages across all batters?
plt.figure(figsize=(8, 6))
plt.hist(df["AVG"], bins=15, color="steelblue", edgecolor="black")
plt.xlabel("Batting Average")
plt.ylabel("Number of Batters")
plt.title("Distribution of Batting Averages - IPL 2025")

plt.show()
```

| --------------------- | ---------------------------------------- |
| `df["4s"] + df["6s"]` | Total boundaries calculate karne ke liye |
| `sort_values()`       | Data sort karne ke liye                  |

```python
# Q2 What is the distribution of batting averages across all batters?
df["Total_boundaries"] = df["4s"] + df["6s"]
top15_bound = df.sort_values("Total_boundaries", ascending=False).head(15)

plt.figure(figsize=(10, 6))
plt.bar(top15_bound["Player Name"], top15_bound["Total_boundaries"], color="olive")
plt.xticks(rotation=60, ha="right")
plt.ylabel("Total Boundaries (4s + 6s)")
plt.title("Top 15 Batters by Total Boundaries - IPL 2025")

plt.show()
```

```python
# Q3 Do batters with more innings generally score more runs
plt.figure(figsize=(8, 6))
plt.scatter(df["Inn"], df["Runs"], alpha=0.7, color="slateblue")
plt.xlabel("Innings")
plt.ylabel("Runs")
plt.title("Innings vs Runs - IPL 2025 Batters")
plt.grid(True, linestyle="--", alpha=0.4)

plt.show()
```

```python
# Q4 Which 10 batters have the highest strike rate (min 150 runs
qualified = df[df["Runs"] >= 150].sort_values("SR", ascending=False).head(10)

plt.figure(figsize=(10, 6))
plt.bar(qualified["Player Name"], qualified["SR"], color="coral")
plt.xticks(rotation=60, ha="right")
plt.ylabel("Strike Rate")
plt.title("Top 10 Batters by Strike Rate (Min 150 Runs) - IPL 2025")

plt.show()
```

```python
# Q5 How does strike rate distribution differ between teams?
import seaborn as sns

plt.figure(figsize=(10, 6))
sns.violinplot(data=df, x="Team", y="SR", inner="quartile")
plt.xticks(rotation=45, ha="right")
plt.title("Strike Rate Distribution by Team - IPL 2025")

plt.show()
```

```python
# Q6 Which batting metrics are strongly correlated with each other?


numeric_cols = ["Runs", "Matches", "Inn", "BF", "SR", "AVG", "4s", "6s"]


for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')

corr = df[numeric_cols].corr()

plt.figure(figsize=(8, 6))
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap of Batting Metrics - IPL 2025")

plt.show()
```

| Function          | Use                              |
| ----------------- | -------------------------------- |
| `pd.to_numeric()` | Data ko numeric me convert karna |
| `errors='coerce'` | Invalid values ko NaN banana     |
| `.corr()`         | Correlation calculate karna      |
| `plt.figure()`    | Graph create karna               |
| `sns.heatmap()`   | Heatmap banana                   |
| `annot=True`      | Values show karna                |
| `cmap="coolwarm"` | Color theme set karna            |
| `fmt=".2f"`       | 2 decimal places dikhana         |
| `plt.title()`     | Title dena                       |
| `plt.show()`      | Graph display karna              |

```python
# Q9 How is the distribution of runs per batter different across teams?
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
sns.boxplot(data=df, x="Team", y="Runs")
plt.xticks(rotation=45, ha="right")
plt.title("Distribution of Runs per Batter by Team - IPL 2025")

plt.show()
```

```python
# Q10 Team-wise boundary contribution – stacked bar
team_boundaries = df.groupby("Team")[["4s", "6s"]].sum()

plt.figure(figsize=(10, 6))
plt.bar(team_boundaries.index, team_boundaries["4s"], label="4s", color="skyblue")
plt.bar(team_boundaries.index, team_boundaries["6s"],
        bottom=team_boundaries["4s"], label="6s", color="navy")

plt.xticks(rotation=45, ha="right")
plt.ylabel("Total Boundaries")
plt.title("Total 4s and 6s by Team - IPL 2025")
plt.legend()
plt.tight_layout()
plt.show()
```

| Function             | Use                   |
| -------------------- | --------------------- |
| `groupby()`          | Team-wise grouping    |
| `sum()`              | Total calculate karna |
| `plt.figure()`       | Graph create karna    |
