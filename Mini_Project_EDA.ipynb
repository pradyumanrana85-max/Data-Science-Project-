#5. Outlier Detection

•⁠  ⁠Detect outliers using a boxplot.
•⁠  ⁠Handle outliers using the IQR or Z-score method.


```python
plt.figure(figsize=(8,5))
sns.boxplot(x=df["release_year"])
plt.title("Boxplot of Release Year")
plt.show()
```

```python
Q1 = df["release_year"].quantile(0.25)
Q3 = df["release_year"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[(df["release_year"] < lower) |
              (df["release_year"] > upper)]

print("Outliers:", len(outliers))
```

#6. Univariate Analysis

Create suitable plots and analyze individual variables.


```python
plt.figure(figsize=(6,4))
sns.countplot(x="type", data=df)
plt.title("Movies vs TV Shows")
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.countplot(y="rating", data=df,
              order=df["rating"].value_counts().index)
plt.title("Rating Distribution")
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.histplot(df["release_year"], bins=30)
plt.title("Release Year Distribution")
plt.show()
```

#7. Bivariate Analysis

Analyze relationships between two variables using appropriate visualizations.



```python
plt.figure(figsize=(12,6))
sns.countplot(x="rating",
              hue="type",
              data=df)

plt.xticks(rotation=45)
plt.show()
```

#8. Correlation Analysis

•⁠  ⁠Create a correlation heatmap.
•⁠  ⁠Identify highly and negatively correlated features.


```python
corr = df.select_dtypes(include=np.number).corr()

plt.figure(figsize=(6,4))
sns.heatmap(corr,
            annot=True,
            cmap="coolwarm")

plt.title("Correlation Heatmap")
plt.show()
```
